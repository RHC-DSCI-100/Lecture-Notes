## Tables

This is one of the most important things we will use in this class. Tables are a way to organize your data for analysis and graphing.

Below are a few of the most common methods used in this class.

```python
# Import all Classes and Functions from the datascience library before you can use it.
from datascience import *
```

```python
# Read a CSV file and convert it into a table in Python memory
cones = Table.read_table('cones.csv')
cones
```

### Show Method

Displays the first n rows of a table. If no n is provided, displays all rows.

```python
cones.show(3)
```

```python
cones.show()
```

### Select Method

Creates a copy of a table with only the selected columns. Each column is the column name as a string or the integer index of the column.

```python
cones.select('Flavor')
```

```python
cones.select('Flavor', 'Price')
```

```python
# The line below causes an error because Flavor is not a name.
cones.select(Flavor, 'Price')
```

### Drop Method

Creates a copy of a table without the specified columns. Each column is the column name as a string or integer index.

```python
cones.drop('Price')
```

```python
cones
```

```python
cones_without_price = cones.drop('Price')
cones_without_price
```

### Where Method

Filters the table for rows where the predicate is true. Predicate should be one of the provided are.<something> functions. Column can be a string column label or an integer index.

```python
cones.where('Flavor', 'chocolate')
```

### Sort Method

Sorts the rows in the table by the values in the column column_name_or_index in ascending order by default. Set descending=True to sort in descending order.

```python
cones.sort('Price')
```

```python
cones.sort('Price', descending=True)
```

```python
cones.sort('Flavor', descending=True)
```
