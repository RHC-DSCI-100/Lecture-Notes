### Standard Deviation ###
A standard measure to help describe the distance a value is from the mean. 

The majority of the data will lie withing 4 standard deviations of the mean. 

***QUESTION: Considering that that standard deviations are in reference to the average, what do you think the sum of all the standard deviations will be?***

```python
sd_table = Table().with_columns('Value', values)
sd_table
```

```python
average_value = np.mean(values)
average_value
```

```python
# Deviation is equal to the data value minus the average value. 

deviations = values - average_value
sd_table = sd_table.with_column('Deviation', deviations)
sd_table
```

```python
sum(deviations)
```

### Variance ###

Since the standard deviation sum is zero, the size of the deviations is calucated by squaring the standard deviation.



```python
sd_table = sd_table.with_column('Squared Deviation', deviations ** 2)
sd_table
```

```python
# Variance of the data: the mean squared deviation from average

variance = np.mean(deviations ** 2)
variance
```

```python
# Standard Deviation (SD): the root mean squared deviation from average
# = square root of the variance

# Brute Force
sd = variance ** 0.5
print(f"Brute: {sd}")

# Math Module
sd = math.sqrt(variance)
print(f" Math: {sd}")

# Numpy Module
sd = np.sqrt(variance)
print(f"Numpy: {sd}")
```

## Standard Units ##

Standardizing the standard deviation tells the number of deviations and direction. 
This is called the z-score and is based on a Normal (symmetric) curve. 
In a Normal distribution, the center is 0 and the Standrad Deviation is 1. 

To standardize, subtract the mean from the data value then divide by the standard deviation.

```python
def standard_units(x):
    """Convert array x to standard units."""
    return (x - np.mean(x)) / np.std(x)
```

```python
ages = births.column('Maternal Age')
ages
```

```python
# What is happening with this code?
ages_standard_units = standard_units(ages)
ages_standard_units
```

```python
np.mean(ages_standard_units), np.std(ages_standard_units)
```
