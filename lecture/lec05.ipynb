## Ranges

You can easily make an array of items like a range by using the make_array() method, but it is not the most efficient way.

```python
make_array(0, 1, 2, 3, 4, 5, 6)
```

### Method arange()

The method `arange()` is from `numpy` and creates a NumPy array filled with evenly spaced values within a specified numerical range. It accepts 4 arguments with only one being required. These arguments are the starting point, stopping point, increment, and `dtype` data type. The stopping point is the only required argument.

`np.arange(start, stop, increment, dtype=None)`

* If `start` is not specified, it defaults to 0.
* If `increment` is not specified, it defaults to 1.
* If `dtype` is not specified, it defaults to None and determines it automatically.


```python
# Range with only stop
np.arange(7)
```

```python
# Range with start and stop
np.arange(5, 11)
```

```python
# Range with start, stop, and increment
np.arange(0, 20, 2)
```

```python
# Range with start, stop, and increment
np.arange(0, 21, 2)
```

```python
# Range with start, stop, and increment
np.arange(0, 1, 0.1)
```

### Method take()

The .take() method on a Table object is used strictly to select specific rows in the order of the provided indices.

```python
# Take index row 0 from warriors table
warriors.take(0)
```

```python
# Take index row 0, 1, 2 from warriors table
warriors.take(0, 1, 2)
```

```python
# Take rows 0-2 using arange() from warriors table
warriors.take(np.arange(3))
```

```python
# Take rows 0-4 using arange() from warriors table then using column $ sum the total
np.sum(warriors.take(np.arange(5)).column('$'))
```

```python
# Take rows 0-4 using arange() from suns table then using column $ sum the total
np.sum(suns.take(np.arange(5)).column('$'))
```

## Creating a Table from Scratch

```python
# Declare an Array of Streets
streets = make_array('Bancroft', 'Durant', 'Channing', 'Haste')
streets
```

```python
# Use the method with_columns() to create a new table with multiple columns
Table().with_columns(
    'Streets', streets,
    'Blocks from campus', np.arange(4)
)
```

## Where Method for Tables

```python
# are.above(x):  Filters rows where the value is strictly greater than a specific value
nba.where('salary', are.above(10000000))
```

```python
# are.between(x, y): Filter rows based on a specific numeric range. By default, it is inclusive of x and exclusive of y (x ≤ value < y)
nba.where('salary', are.between(10000000, 20000000))
```

```python
# are.containing(S): Filters rows where a specified column's string contains the substring S
nba.where('position', are.containing('G'))
```
