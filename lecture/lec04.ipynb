There are two built-in numerical data types this class will be using, integer and float.

* **`int` (Integer)** is a whole number, positive or negative, without decimals, of unlimited length with respect to the amount of memory you have.

Float can also be scientific numbers with an "e" to indicate the power of 10.

### Mixed Data Types

In Python numeric expressions, mixing int and float types automatically triggers implicit type conversion (coercion), which always upgrades the result to a float.

```python
10 ** 0.5
```

### Float Significant Digits

In Python, the built-in float type is implemented as a 64-bit double-precision binary number conforming to the IEEE 754 standard. This means it provides 53 bits of precision, which translates to roughly 15 to 17 significant decimal digits.

```python
75892745.215489247589274985712
```

## Collection Data Types

Python has four built-in collection data types used to store groups of data: `list`, `tuple`, `set`, and `dict` (dictionary). Each type has unique characteristics regarding order, mutability, and how elements are accessed. We will be using the list, tuple, and ocassionally dictionary.

* **`List`**: An ordered, mutable sequence that allows duplicate members. It is the most versatile and frequently used collection type
* **`Tuple`**: An ordered, immutable sequence that allows duplicate members. Once created, you cannot change, add, or remove its elements.
* **`Dictionary`**: A collection of key-value pairs where keys must be unique and immutable, while values can be of any data type.


```python
# Tuple: Comma separated list of elements, can also use (1, 234, 456)
1, 234, 456
```

```python
# List: Comma separted list of elements that uses the square brackets to denote it is a list
[1, 234, 456]
```

```python
# Dictionary: Elements use key value pairs. Key is separated by a colon and each pair is separated by a comma.
# Dictionary uses curly brackets with the pairs to denote it is a dictionary.
{ 'key1': 1, 'key2': 234, 'key3': 456 }
```

## Arrays

* An array is a container that holds elements
* The index of the first element starts at zero
* Arrays can have operations performed on them.


### Create an Array

The make_array() function is a simplified utility used to quickly create NumPy arrays. It is highly useful for building arrays manually, passing data into functions, or adding new columns to a Table.

Note all arguements in the array should be the same data type.

```python
first_four = make_array(1, 2, 3, 4)
```

```python
first_four
```

You can perform math operations on the whole array instantly. If you multiply, add, or divide a make_array() object by a number, it performs that operation on every element automatically.

```python
first_four ** 2
```

```python
first_four + 1
```

```python
first_four / 2
```

## Columns of Tables are Arrays ##

```python
nba = Table.read_table('nba_salaries.csv').where('season', 2020).drop('season')
nba.show(5)
```

```python
warriors = nba.where('team', 'Golden State Warriors')
warriors.show(5)
```

```python
warriors.select('salary')
```

```python
type(warriors.select('salary'))
```

### Table: Column Method

The singular .column() method is used to fetch an array of data based on the provided column.

```python
warriors.column('salary')
```

```python
type(warriors.column('salary'))
```

### Table: Average Method

```python
np.average(warriors.column('salary'))
```

```python
suns = nba.where('team', 'Phoenix Suns')
```

```python
np.average(warriors.column('salary')) - np.average(suns.column('salary'))
```

### Table: Sum Method

```python
np.sum(warriors.column('salary')) - np.sum(suns.column('salary'))
```
